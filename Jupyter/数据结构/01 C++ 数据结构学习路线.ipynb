# 数据结构学习流程

> 目标：以 `C/C++` 为实现语言，把“概念理解 + 手写实现 + 复杂度分析 + 经典题应用”串成一条完整主线。  
> 依据：`课程.md` 中的完整播放列表。  
> 原则：主线尽量短，但关键知识不能断层；先学“能搭框架的内容”，再补“考研/面试高频专题”。

---

但对于初学者，第一轮只要覆盖“图的表示+图的遍历”，即可掌握图的基本代码逻辑（这也是面试超高频的手写点）。复杂的算法可待第二轮强化时再重点攻克。如果第一轮强行学完，极易出现挫败感和遗忘。

**目标**：建立“平均意义上的高效查找”认知。  
**来源**：`王道考研`

### 主线视频

- 7.5.1 散列表的基本概念
- 7.5.2 散列函数的构造
- 7.5.3_1 拉链法
- 7.5.3_2 开放定址法
- 7.5.4 散列查找的性能分析

### 这一阶段必须会

- 哈希表为什么查找快。
- 冲突为什么不可避免。
- 拉链法和开放定址法的区别。
- 装填因子为什么重要。

### 输出要求

- 手写一个简单哈希表。
- 至少实现一种冲突处理策略。

## 阶段十三：堆 ｜ ~0.7h

**目标**：把堆当成“完全二叉树上的优先队列结构”来理解，而不是只把它当排序附属品。  
**来源**：`王道考研`

### 主线视频

- 8.4.2_1 堆排序
- 8.4.2_2 堆的插入删除

### 建议搭配

如果你觉得只看这两节不够顺，可以回看：

- `Abdul Bari 课程一` 2.6.3 堆 - 堆排序 - 堆化 - 优先级队列

### 这一阶段必须会

- 大根堆 / 小根堆性质。
- 上滤、下滤、建堆。
- 堆排序为什么是 `O(n log n)`。
- 堆为什么适合做优先队列。
