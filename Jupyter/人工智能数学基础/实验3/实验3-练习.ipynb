# 🧮 基与秩综合练习题

### 📖 预备知识：QR 分解、正交投影与基下坐标

**1. QR 分解（QR Decomposition）**

QR 分解是将一个矩阵分解为两个矩阵的乘积：

$$
A=QR
$$

其中：

- $Q$：列向量相互正交且长度为 1，即 $Q^TQ=I$。
- $R$：上三角矩阵。

QR 分解的一个重要作用，是帮助我们找到列空间的一组**标准正交基**。

如果矩阵 $A$ 的秩为 $r$，可以从其列空间中选取 $r$ 个线性无关的向量，构成矩阵 $B$，再进行 QR 分解：

$$
B=Q_rR
$$

其中 $Q_r=[q_1,\ldots,q_r]$ 的列向量构成 $\operatorname{Col}(A)$ 的一组标准正交基，满足：

$$
\boxed{Q_r^TQ_r=I_r}
$$

在 Python 中，可以使用 `np.linalg.qr(B, mode='reduced')` 进行计算。

**2. 正交投影（Orthogonal Projection）**

假设有一个向量 $y\in\mathbb R^5$，以及一个由若干向量张成的子空间 $W$。如果 $y$ 不在 $W$ 中，我们希望找到 $W$ 中距离 $y$ 最近的向量 $\hat y$，这个向量就称为 $y$ 在 $W$ 上的**正交投影**。

正交投影有两个重要特点：

- $\hat y\in W$，即投影结果属于目标子空间。
- $r=y-\hat y$ 与子空间 $W$ 中所有向量正交。

如果 $W$ 的一组标准正交基为 $q_1,\ldots,q_r$，组成矩阵 $Q_r=[q_1,\ldots,q_r]$，则正交投影可以计算为：

$$
\boxed{\hat y=Q_rQ_r^Ty}
$$

其中，$Q_r^Ty$ 计算 $y$ 在各个正交基方向上的分量，再通过 $Q_r$ 将这些分量组合成投影向量。

**3. 基下坐标（Coordinates Relative to a Basis）**

同一个向量，在不同的基下可以有不同的坐标表示。

假设子空间 $W$ 的一组基为

$$
B=\{b_1,b_2,b_3\}
$$

如果向量 $y\in W$，那么它可以唯一表示为

$$
y=\alpha_1b_1+\alpha_2b_2+\alpha_3b_3
$$

其中 $\alpha_1,\alpha_2,\alpha_3$ 就是向量 $y$ 在基 $B$ 下的坐标，记为

$$
\boxed{
[y]_B=
\begin{bmatrix}
\alpha_1\\
\alpha_2\\
\alpha_3
\end{bmatrix}
}
$$

例如，若 $y=2b_1-3b_2+b_3$，则

$$
[y]_B=
\begin{bmatrix}
2\\
-3\\
1
\end{bmatrix}
$$

注意：**$y$ 是向量本身，而 $[y]_B$ 是它在指定基下的坐标。** 两者描述的是同一个向量，但表达方式不同。

## 📘 一、题目背景

我们希望通过一个 5×4 的矩阵来理解 **秩（rank）**、**基（basis）** 与 **线性相关性（linear dependence）** 之间的关系，  
并用 Python 进行验证与计算。

## 🧩 二、数据定义

已知四个列向量 $$x_1,x_2,x_3,x_4\in\mathbb{R}^5$$，构造矩阵：

$$
x_1 = 
\begin{bmatrix} 1 \\ 2 \\ -1 \\ -1 \\ -1 \end{bmatrix},
\quad
x_2 = 
\begin{bmatrix} 2 \\ -1 \\ 1 \\ 2 \\ -2 \end{bmatrix},
\quad
x_3 = 
\begin{bmatrix} 3 \\ -4 \\ 3 \\ 5 \\ -3 \end{bmatrix},
\quad
x_4 = 
\begin{bmatrix} -1 \\ 8 \\ -5 \\ -6 \\ 1 \end{bmatrix}
$$

$$
A = [x_1 \; x_2 \; x_3 \; x_4]
$$

## 🧮 三、任务要求

1. **求矩阵的秩**  
   - 使用 `np.linalg.matrix_rank(A)`  
   - 理解“秩”代表矩阵中**最大线性无关列数**。

2. **确定列空间的基**  
   - 用 `sympy.Matrix(A).rref()` 求行最简形（RREF）。  
   - 记录主元列索引（pivot columns）。  
   - 从 $x_1,x_2,x_3,x_4$ 中选出相应列，构成矩阵的**列空间基**。

3. **验证线性关系**  
   - 假设 $x_3 = a x_1 + b x_2 + c x_4$，  
     用 `np.linalg.lstsq(A, x3)` 或选取主元列矩阵解出 $[a,b,c]$。  
   - 判断哪些列可以被其它列线性表示。

4. **构造正交基（QR分解）**  
   - 用 `np.linalg.qr(A)` 得到 \(Q, R\)。  
   - 取 $Q$ 的前 `rank(A)` 列组成正交基 $Q_r$。  
   - 验证 $Q_r^T Q_r = I$。

5. **计算正交投影**  
   - 给定向量 $y = [1, 0, 2, -1, 0]^T$，  
     求其在 $\mathrm{Col}(A)$ 上的正交投影：  
     $\hat{y} = Q_r Q_r^T y$。  
   - 计算残差 $r = y - \hat{y}$，并验证 $Q_r^T r = 0$。

6. 求坐标表示

    - 设列空间的一组基为 \(B=\{b_1,b_2,b_3\}\)。
    - 将第5问得到的正交投影 \(\hat y\) 表示为这组基的线性组合：
    $$
    \hat y=\alpha_1b_1+\alpha_2b_2+\alpha_3b_3
    $$
    - 求坐标向量 $$[\hat y]_B=\alpha$$。
    - 可用 np.linalg.lstsq(B, y_hat, rcond=None) 实现。

## 🧠 四、思考提示

| 步骤 | 思考方向 | Python 提示 |
|------|------------|-------------|
| 1️⃣ 秩 | 代表“空间维数” | `matrix_rank(A)` |
| 2️⃣ 基 | 选出“最小无关组” | `sp.Matrix(A).rref()` |
| 3️⃣ 线性关系 | 判断多余列 | 解 `A[:,piv] * α ≈ x3` |
| 4️⃣ 正交化 | 保留方向但使独立 | `Q, R = qr(A)` |
| 5️⃣ 投影 | 找“影子” | `y_hat = Qr @ Qr.T @ y` |
| 6️⃣ 坐标 | 在基下的表示 | `alpha, *_ = lstsq(B, y)` |

```python
import numpy as np
from numpy.linalg import matrix_rank, lstsq, qr
import sympy as sp


# 构造矩阵 A
x1 = np.array([ 1,  2, -1, -1, -1.], dtype=float)
x2 = np.array([ 2, -1,  1,  2, -2.], dtype=float)
x3 = np.array([ 3, -4,  3,  5, -3.], dtype=float)
x4 = np.array([-1,  8, -5, -6,  1.], dtype=float)
A  = np.c_[x1, x2, x3, x4]  # 5x4

# 1) 秩
rank_A = matrix_rank(A)
print("rank(A) =", rank_A)

# 2) 主元列与列空间基
M = sp.Matrix(A.astype(int))
RREF, piv = M.rref()  # piv 是主元列索引（从0起）
print("pivot columns (0-based):", piv)
B = A[:, piv]
print("RREF(A) =")
sp.pprint(RREF)
print("B =")
print(B)

# 3) 线性关系：在所选基上表示 x3
coef_3, *_ = lstsq(B, x3, rcond=None)
print("x3 in B:", coef_3)
print("B @ coef_3 == x3:", np.allclose(B @ coef_3, x3))
print("x1 = 2*x2 - x3:", np.allclose(x1, 2 * x2 - x3))
print("x2 = (x1 + x3)/2:", np.allclose(x2, (x1 + x3) / 2))
print("rank([x1, x2, x3]) =", matrix_rank(A[:, :3]))

# 4) 正交基（QR）
Q, R = qr(B, mode='reduced')
Qr = Q[:, :rank_A]
print("Qr =")
print(Qr)
print("R =")
print(R)
print("Qr.T @ Qr =")
print(Qr.T @ Qr)
print("Qr.T @ Qr == I:", np.allclose(Qr.T @ Qr, np.eye(rank_A)))
print("Qr @ R == B:", np.allclose(Qr @ R, B))

# 5) 投影 y 到 Col(A)
y = np.array([1., 0., 2., -1., 0.])
y_hat = Qr @ (Qr.T @ y)
resid = y - y_hat
print("y_hat =", y_hat)
print("resid =", resid)
print("Qr.T @ resid =", Qr.T @ resid)
print("Qr.T @ resid == 0:", np.allclose(Qr.T @ resid, np.zeros(rank_A)))

# 6) y_hat 在所选基 B 下的坐标
alpha, *_ = lstsq(B, y_hat, rcond=None)
print("[y_hat]_B =", alpha)
print("B @ alpha == y_hat:", np.allclose(B @ alpha, y_hat))

# 附：验证课堂给出的关系 x3 = -x1 + 2x2（若主元列是 {x1,x2,x4}）
coef_12, *_ = lstsq(np.c_[x1, x2], x3, rcond=None)
print("x3 in {x1,x2}:", coef_12, " -> check:",
      np.allclose(x3, np.c_[x1, x2] @ coef_12))
```

# 🔢 线性映射与基变换练习题

## 📖 预备知识：基下坐标与基变换矩阵

**1. 为什么 $[x]_B=P^{-1}x$？**

在 $\mathbb R^3$ 中，向量通常使用标准基表示。例如：

$$
x=\begin{bmatrix}2\\-1\\1\end{bmatrix}
$$

现在换一组基：

$$
B=\{b_1,b_2,b_3\}
$$

同一个向量 $x$ 可以表示为：

$$
x=\alpha_1b_1+\alpha_2b_2+\alpha_3b_3
$$

其中 $\alpha_1,\alpha_2,\alpha_3$ 就是 $x$ 在新基 $B$ 下的坐标：

$$
[x]_B=\begin{bmatrix}\alpha_1\\\alpha_2\\\alpha_3\end{bmatrix}
$$

把三个基向量按列组成矩阵：

$$
P=[b_1\ b_2\ b_3]
$$

那么上面的线性组合就可以写成：

$$
x=P[x]_B
$$

因为 $B$ 是一组基，所以 $P$ 可逆，两边乘以 $P^{-1}$：

$$
\boxed{[x]_B=P^{-1}x}
$$

因此：

- $P$：将新基 $B$ 下的坐标转换为标准基下的坐标。
- $P^{-1}$：将标准基下的坐标转换为新基 $B$ 下的坐标。

**2. 为什么 $A_B=P^{-1}AP$？**

假设线性映射为：

$$
T(x)=Ax
$$

其中 $A$ 是线性映射在标准基下的矩阵。

现在我们希望找到它在新基 $B$ 下的矩阵 $A_B$，使得：

$$
[T(x)]_B=A_B[x]_B
$$

整个过程可以分成三步：

**第一步：将新基坐标转换为标准基坐标**

$$
x=P[x]_B
$$

**第二步：在标准基下进行线性变换**

$$
T(x)=Ax=AP[x]_B
$$

**第三步：将变换结果转换回新基坐标**

$$
[T(x)]_B=P^{-1}T(x)=P^{-1}AP[x]_B
$$

因此：

$$
\boxed{A_B=P^{-1}AP}
$$

这个公式可以理解为：

$$
\underbrace{P^{-1}}_{\text{转回新基}}
\underbrace{A}_{\text{执行线性变换}}
\underbrace{P}_{\text{转到标准基}}
$$

注意矩阵乘法从右往左执行。

**3. 关键理解**

$A$ 和 $A_B$ 虽然数值可能不同，但它们表示的是**同一个线性映射 $T$**，区别只在于使用了不同的基。

就像同一个位置可以用不同的坐标系描述一样，改变坐标表示，并不意味着实际的向量或线性变换发生了改变。

> ✅ **结论：**

> - 基变换是“换语言不换规律”的过程；
> - 数学上，$A$ 与 $A_B$ 描述的是**同一个线性规律 T**。




```python
import sympy as sp

# 数据：矩阵A，新基B的列向量b1,b2,b3，基矩阵P，向量x
A = sp.Matrix([[2,1,-1],
               [0,1, 3],
               [0,0, 2]])
b1 = sp.Matrix([1,1,0])
b2 = sp.Matrix([1,0,1])
b3 = sp.Matrix([0,1,1])
P  = sp.Matrix.hstack(b1,b2,b3)
x  = sp.Matrix([2,-1,1])

# Q1 秩
rank_A = A.rank()

# Q2 可逆性（行列式）
det_A = A.det()
is_invertible = det_A != 0

# Q4 像：T(x)=A x
Tx = A * x

# Q5 坐标： [x]_B = P^{-1} x  （LUsolve 比直接求逆更稳健）
x_B = P.LUsolve(x)


# Q6 一致性： [T(x)]_B ?= A_B [x]_B
Tx_B = P.LUsolve(Tx)

# Q7 新基下矩阵：A_B = P^{-1} A P
A_B = P.LUsolve(A * P)

# 打印关键结果
print("rank(A) =", rank_A)
print("det(A) =", det_A)
print("T 可逆（双射）:", is_invertible)
print("秩 = dim(Im(T)) =", rank_A)
print("T(x) =")
sp.pprint(Tx)
print("[x]_B =")
sp.pprint(x_B)
print("[T(x)]_B =")
sp.pprint(Tx_B)
print("A_B =")
sp.pprint(A_B)
print("A_B * [x]_B =")
sp.pprint(A_B * x_B)
print("[T(x)]_B == A_B * [x]_B:", Tx_B == A_B * x_B)
print("A * P == P * A_B:", A * P == P * A_B)
print("A 与 A_B 表示同一线性映射，使用不同的基。")
```


# 🤖  应用题

## 一、背景

在机器学习中，一个线性层就是一个线性映射：
$$
T(x)=A x
$$
当我们**更换输入特征的基**（重新编码特征）时，线性层的**矩阵表达会随之改变**，  
但**几何作用不变**。本题让你通过计算与代码体会这一原理。

我们使用如下线性层（从 $\mathbb{R}^4$ 到 $\mathbb{R}^3$）：
$$
A=
\begin{bmatrix}
1 & 0 & 2 & -1\\
0 & 1 & 1 &  1\\
1 & 1 & 3 &  0
\end{bmatrix}.
$$

定义新的输入基：
$$
b_1=\begin{bmatrix}1\\0\\0\\0\end{bmatrix},\;
b_2=\begin{bmatrix}1\\1\\0\\0\end{bmatrix},\;
b_3=\begin{bmatrix}0\\1\\1\\0\end{bmatrix},\;
b_4=\begin{bmatrix}0\\0\\1\\1\end{bmatrix}.
$$

基矩阵为：
$$
P=[b_1\;b_2\;b_3\;b_4].
$$

---

## 二、问题与提示

### 1️⃣ 计算 $\operatorname{rank}(A)$  
它表示该层“**有效利用的独立输入方向数**”。

**提示：**
- 用 `A.rank()` 求秩；
- 若某一行能由其他行线性表示，则输出落在低维空间；
- $\operatorname{rank}(A)\le \min(3,4)=3$。

---

### 2️⃣ 判断是否有维度压缩  
若 $\operatorname{rank}(A)<4$，则有 $(4-\operatorname{rank}(A))$ 个输入方向被“压平”。

**提示：**
- 线性层的输出维度 = $\operatorname{rank}(A)$；
- 维度差值即丢失信息量。

---

### 3️⃣ 在新基 $B$ 下的矩阵  
只更换输入基时：
$$
A_B = A P.
$$
这表示：**先把新坐标还原为旧坐标（乘 $P$），再通过线性变换 $A$。**

---

### 4️⃣ 验证 $T(x)=A x=A_B[x]_B$  
取 $x=\begin{bmatrix}2\\1\\0\\1\end{bmatrix}$。

**提示：**
- 用 `P.LUsolve(x)` 求 $[x]_B=P^{-1}x$；
- 比较两边是否相等；
- 若相等，则说明换基后映射保持一致。

---

### 5️⃣ 验证输出是否落在二维平面  
若某行是其它行之和（如 $r_3=r_1+r_2$），  
则对所有 $x$，有：
$$
(Ax)_3=(Ax)_1+(Ax)_2,
$$
输出落在平面
$$
\{y\in\mathbb{R}^3 \mid y_3=y_1+y_2\}.
$$

---



```python
import sympy as sp

# ==== 0. 数据设置 ====
A = sp.Matrix([
    [1, 0, 2, -1],
    [0, 1, 1,  1],
    [1, 1, 3,  0]
])
b1 = sp.Matrix([1,0,0,0])
b2 = sp.Matrix([1,1,0,0])
b3 = sp.Matrix([0,1,1,0])
b4 = sp.Matrix([0,0,1,1])
P  = sp.Matrix.hstack(b1,b2,b3,b4)
x  = sp.Matrix([2,1,0,1])

print("A ="); sp.pprint(A)
print("\nP ="); sp.pprint(P)
print("\nx =", x.T)

# ==== 1. 秩与“压缩维度” ====
rank_A = A.rank()
compressed_dim = A.cols - rank_A
print("\nrank(A) =", rank_A)
print("dim(Im(T)) =", rank_A)
print("dim(Ker(T)) =", compressed_dim)
print("存在维度压缩:", compressed_dim > 0)

# ==== 3. 新基下的矩阵 A_B ====
A_B = A * P
print("\nA_B = A * P =")
sp.pprint(A_B)

# ==== 4. 坐标与一致性验证 ====
Tx = A * x
x_B = P.LUsolve(x)        # [x]_B = P^{-1} x
Tx_via_B = A_B * x_B      # A_B [x]_B
print("\nT(x) =")
sp.pprint(Tx)
print("[x]_B =")
sp.pprint(x_B)
print("A_B * [x]_B =")
sp.pprint(Tx_via_B)
print("A * x == A_B * [x]_B:", Tx == Tx_via_B)
print("(Tx)_3 == (Tx)_1 + (Tx)_2:", Tx[2] == Tx[0] + Tx[1])

# ==== 4. 说明“输出落在二维平面上”的性质 ====
# 因为 row3 = row1 + row2，所以 (Ax)_3 = (Ax)_1 + (Ax)_2 对任意 x 成立
r1, r2, r3 = A.row(0), A.row(1), A.row(2)
print("\n行关系检查： row3 == row1 + row2 ? ->", (r3 == r1 + r2))

# 用符号变量验证 (Ax)_3 - (Ax)_1 - (Ax)_2 == 0
x_sym = sp.symbols('x1 x2 x3 x4')
x_vec = sp.Matrix(x_sym)
expr = (A * x_vec)[2] - (A * x_vec)[0] - (A * x_vec)[1]
print("符号验证 (Ax)_3 - (Ax)_1 - (Ax)_2 = ", sp.simplify(expr))
```
